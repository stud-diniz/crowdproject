```python
# gets lists for each of the runs, and diff ones, if you change somthing in the data, copy from flow over time, delete the first line and call it here 
# shouldnt be needed anymore 
from collections import defaultdict

groups = defaultdict(list)

filename = r"C:\Users\vms3\Downloads\flow_over_time.csv"

# --- READ FILE ---
with open(filename) as f:
    for line in f:
        parts = line.strip().split(',')

        # skip empty/bad lines
        if not parts or parts[-1].strip() == '':
            continue

        try:
            sim = parts[0]
            value = int(parts[-1].strip())
            groups[sim].append(value)
        except ValueError:
            continue

# --- PRINT ORIGINAL LISTS ---
for sim, values in groups.items():
    print(f"\nsim{sim} = [")
    for v in values:
        print(f"    {v},")
    print("]")

# --- COMPUTE DIFFERENCES ---
diff_groups = {}

for sim, values in groups.items():
    diff = [0]  # first value stays 0
    for i in range(1, len(values)):
        diff.append(values[i] - values[i-1])
    diff_groups[sim] = diff

# --- PRINT DIFFERENCE LISTS ---
for sim, values in diff_groups.items():
    print(f"\nsim{sim}d = [")
    for v in values:
        print(f"    {v},")
    print("]")

```
